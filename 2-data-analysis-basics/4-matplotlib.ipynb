#### Basics of matplotlib.pyplot

```python
import matplotlib.pyplot as plt
```

##### Line PLot

```python
# plotting line plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
plt.plot(x, y)
# description about the plot
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# Customizing out plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 9, 16, 25]

plt.plot(x, y, color="orange", marker="o", linestyle="--", linewidth=1, markersize=8)
plt.grid(True)
```

##### Subplot (multiple plot)

```python
# multiple plot using subplot
# So we will have multiple plot either in a single row or single column based on our configuration
x1 = [1, 2, 3, 4, 5]
y1 = [2, 5, 6, 12, 20]
x2 = [1, 2, 3, 4, 5]
y2 = [2, 4, 9, 16, 25]

# here we are making the space in which our plot will be displayed. height=10 and width=5
plt.figure(figsize=(10, 5))
# ============== First plot ==============
# we will have 1 row and 2 columns in our plot
# 1 row and 2 columns ,third parameter 1 means the following plot will be in 1st column
plt.subplot(1, 2, 1)
plt.plot(x1, y1)
plt.title("Plot 1")

# ============== Second plot ==============
# 1 row and 2 columns ,third parameter 2 means the following plot will be in 2nd column
plt.subplot(1, 2, 2)
plt.plot(x2, y2, color="red")
plt.title("Plot 2")
plt.show()
```

#### Bar plot

```python
# bar plot is generally used for categorical data and it shows the frequency of each category
categories = ["A", "B", "C", "D"]
values = [10, 20, 30, 40]
plt.bar(categories, values, edgecolor="yellow")
```

#### Histogram

```python
# we use continuous data to plot histogram
x = [5, 1, 3, 3, 2, 6, 1, 1, 4, 5, 3]

# we can have multiple histogram in a single plot
plt.hist(x, bins=5, color="teal")
plt.show()
```

The difference between bar plot and histogram is, Barplot are discrete and used with categorical data so we have to provide value that says the frequency of that category.
In Histogram we just have to the continuous value and give the bin size. each bin size will have frequency of that value in that bin range

#### Scatter Plot

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 7, 5, 9]
plt.scatter(x, y, color="red", marker="*")
plt.show()
```

#### Pie chart

```python
category = ["A", "B", "C", "D"]
values = [10, 20, 30, 40]
colors = ["red", "blue", "green", "orange"]
plt.pie(
    values,
    labels=category,
    colors=colors,
    explode=(0.0, 0.0, 0.05, 0.0),
    autopct="%1.1f%%",
    shadow=True,
)
plt.show()
```
