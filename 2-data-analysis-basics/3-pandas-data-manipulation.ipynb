### Some important data manipulation used in pandas

```python
import pandas as pd
from IPython.display import display
```

```python
# read the data.csv file
df = pd.read_csv("data.csv")
# print the firt 5 entry of the dataframe
display(df.head(5))
```

```python
# statistical analysis of the dataframe
display(df.describe())
```

```python
# print how many null or empty values are present in the dataframe for the respective column
df["Sales"].isnull().sum()
# ? check if any column is there that has any missing values
display(df.isnull().any())
# We can see
```

```python
# Handling Missing Values

# drop all the rows that has any missing values in any of its columns
# df.dropna(inplace=True)

# filling the null values in a column with mean of that column
df["Value"] = df["Value"].fillna(df["Value"].mean())
df["Sales"] = df["Sales"].fillna(df["Sales"].mean())
df.isnull().sum()
```

```python
# Rename old column name with new column name
# First Print all the column name
print(df.columns)
# rename Sales to TotalSales
df.rename(columns={"Sales": "TotalSales"}, inplace=True)
# Print after renaming column
print(df.columns)
```

```python
# * Applying function to change or add new column data
print(df["Value"].head())
df["Value"] = df["Value"].apply(lambda x: x * 2)
print("=======After applying lambda function ======= : \n", df["Value"].head())
```

```python
# * Data Aggreagting and Grouping
# Find the total sum of values for each product
display(df.groupby("Product")["Value"].sum())
# we can group by based on 2 or many categorical columns
display(df.groupby(["Product", "Region"])["Value"].sum())
df.head()
```

```python
# Merging two dataframe rowise and column wise
df1 = pd.DataFrame({"col1": [1, 2, 3], "col2": [4, 5, 6]})
df2 = pd.DataFrame({"col1": [1, 2, 4], "col3": [42, 53, 64]})
# Merging Inner join
print(pd.merge(df1, df2, on="col1", how="inner"))
print("============================>")
# Merging Outer Join
print(pd.merge(df1, df2, on="col1", how="outer"))
print("============================>")
# Merging left Join
print(pd.merge(df1, df2, on="col1", how="left"))
```
