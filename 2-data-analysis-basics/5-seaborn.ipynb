## Basic Seaborn plotting visualization

Seaborn is high-level visualization library that uses matplotlib under the hood and able to create much better and interactive visual plot.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("whitegrid")
```

```python
# loading the in-build dataset
tips_dataset = sns.load_dataset("tips")
tips_dataset.head()
```

#### Scatter Plot

```python
# Scatter plot
# hue will color based on some category
sns.scatterplot(x="total_bill", y="tip", data=tips_dataset, color="green", hue="sex")
plt.xlabel("total_bill")
plt.ylabel("tip")
plt.title("Scatter plot - How much tips and total bill are related")
```

#### Line Plot

```python
# here we are comparing how total_bill changes based on size.
# so total_bill is the output so we have placed in y axis, and independent feature in x axis i.e size.
# NOTE: Based on size total_bill will change, not based on total_bill size will change
sns.lineplot(x="size", y="total_bill", data=tips_dataset)
plt.xlabel("size")
plt.ylabel("total_bill")
plt.title("Line plot - How total_bill changes based on Size")
plt.show()
```

#### Categorical Plot - (Bar Plot, Box plot, violin plot)

```python
sns.barplot(x="day", y="total_bill", data=tips_dataset)
plt.title("Bar Plot - How total_bill changes based on Day")
plt.show()
```

```python
# Box plot (Useful for identifying outliers)
sns.boxplot(x="day", y="total_bill", data=tips_dataset)
plt.show()
```

```python
# showing the maximum total bill spend on day wise
display(tips_dataset.groupby("day")["total_bill"].max())
```

```python
sns.violinplot(x="day", y="total_bill", data=tips_dataset)
plt.show()
```

### Histogram - To see the distribution of data

```python
sns.histplot(x="total_bill", data=tips_dataset, kde=True, color="purple")
# We can look into the distribution and say most people spend in between 10 to 20 dollars
plt.show()
```

#### Pair Plot (Help to plot different graph for diff combination of features)

```python
sns.pairplot(tips_dataset)
plt.show()
```

```python
# correlation is used to find the relation between two variables.
# since we divide by standard deviation it is normalized [-1,1]
# this can be used for heatmap
corr = tips_dataset.corr(numeric_only=True)
sns.heatmap(corr, cmap="coolwarm", annot=True, square=True)
```

```python
# we can use aggregation function straight into seaborn plot
sns.barplot(x="sex", y="total_bill", data=tips_dataset, estimator="sum")
plt.show()
print(tips_dataset.groupby("sex")["total_bill"].sum())
```
