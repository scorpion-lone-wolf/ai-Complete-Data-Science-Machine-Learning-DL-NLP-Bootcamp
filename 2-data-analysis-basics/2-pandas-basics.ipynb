```python
import pandas as pd
from IPython.display import display
```

```python
# Series => Series are 1 dim array like Object which is similar to Column in an table
data_1 = [1, 2, 3, 4, 5]
pd.Series(data=data_1)
```

```python
# Creating a Series from dictionary element
# Since dictionary has key and value => Key will be map to index (either number or label if string type) and value will be the actual value of a Series
my_dic = dict({101: 1, 102: 2, 103: 3, 104: 4, 105: 5})
print("Creating Series from dictionary : \n", pd.Series(my_dic))
```

```python
my_dict = {
    "Name": ["Rahul", "Pursotam", "undertaker"],
    "Age": [20, 21, 22],
    "City": ["Delhi", "Kolkata", "Chennai"],
}
# creating df from dictionary
df = pd.DataFrame(
    my_dict, index=[1, 2, 3]
)  # here index=[] will be new label for the rows
display(df)
```

```python
# Data manipulation using pandas

# Adding new column i,e salary to this df
df["Salary"] = df["Age"] * 10000
display(df)
#  droping the salary column
df.drop(
    "Salary", axis=1, inplace=True
)  # inplace will make permanent change in df instead of returning new df with droped column
print("dataframe after dropping salary column")
display(df)
```

```python
# get the Statistical summary of the dataframe
df.describe()
```

#### Some basic data manipulation in pandas

```python
# DataFrame => They are collection of Pandas Series. So they represent both rows and columns of a table
# read the csv file
sales_df = pd.read_csv("sales_data.csv")
# print the first 5 entry of the dataframe
display("The first 5 entry of the sales data :", sales_df.head(5))
```

```python
# Accessing data from dataframe
sales_df["Product Category"].head(4)
```

```python
# indexing in dataframe
# loc = it uses label based indexing i.e based on label name it can access the data .loc[row,col]
print(
    sales_df.loc[0, "Product Category"]
)  # here we use 0 as row because by default the lable for row in df is 0,1,2 and so on
# iloc is used for accessing using index position like 0,1,2,3 and so on
print(sales_df.iloc[0, 2])  # here the data is access based on pure index value
```
