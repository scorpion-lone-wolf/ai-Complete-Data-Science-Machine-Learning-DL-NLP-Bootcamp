```python
# Reference Counting
# It is a mechanism that keeps track of the number of references to a particular object in memory.
# When the reference count reaches zero, the object is automatically deallocated.
import sys

a = []
print(sys.getrefcount(a))
# output will be 2 = one from a , and another from getrefcount itself
```

```python
b = a
print(sys.getrefcount(a))  # 3
del b
print(sys.getrefcount(a))  # 2
```
