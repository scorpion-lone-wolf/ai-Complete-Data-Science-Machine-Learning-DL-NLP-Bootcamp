# Nominal Encoding Using One-Hot Encoding

Nominal data is categorical data that has no natural order.

Examples:

- City
- Gender
- Country
- Color

In this notebook, we will convert the `City` column into numbers using **One-Hot Encoding**.

## Why Encoding Is Needed

Most machine learning models cannot directly understand text values like `Delhi`, `Mumbai`, or `Kolkata`.

So we convert these text categories into numeric columns with values `0` and `1`.

## What We Will Learn

1. Create a simple dataset
2. Check categorical values
3. Apply One-Hot Encoding
4. Create encoded column names
5. Convert encoded output into a DataFrame
6. Join encoded data with the original data
7. Handle unseen categories


## 1. Import Required Libraries

We import the required libraries first.

- `pandas` is used to create and handle the DataFrame.
- `OneHotEncoder` is used to encode categorical columns.


```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

## 2. Create a Simple Dataset

Here we create a small customer dataset.

The dataset contains:

- `Customer`: customer name
- `City`: categorical column
- `Gender`: categorical column
- `Age`: numeric column

For this notebook, we will focus on the `City` column.


```python
#  Creating a simple dataset
import pandas as pd

data = pd.DataFrame(
    {
        "Customer": ["Rahul", "Amit", "Priya", "John", "Divya"],
        "City": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai"],
        "Gender": ["Male", "Male", "Female", "Male", "Female"],
        "Age": [23, 30, 27, 35, 25],
    }
)

data.dtypes
```

### Dataset Observation

The `City` column has text values.

Text values must be converted into numbers before they are used in most machine learning models.


## 3. Check Unique Values in City Column

Before applying encoding, we should check the unique categories in the column.

This helps us understand how many new columns One-Hot Encoding will create.


```python
#  Lets work on city column
data["City"].unique()
```

### Unique Value Observation

Each unique city will become a separate encoded column.

For example, if there are 3 unique cities, One-Hot Encoding creates 3 city columns.


## 4. Apply One-Hot Encoding

Now we create an object of `OneHotEncoder` and apply it on the `City` column.

`fit_transform()` does two things:

1. `fit` learns all unique city names from the data.
2. `transform` converts those city names into numeric values.

We also use `handle_unknown="ignore"`.

This means if a new city appears later, the encoder will not throw an error. It will encode that unknown city as all zeros.


```python
#  lets create an object of OneHotEncoder
ohe = OneHotEncoder(handle_unknown="ignore")
encoded = ohe.fit_transform(data[["City"]])
encoded.toarray()
```

### Encoding Observation

The encoded output contains only `0` and `1`.

For each row:

- `1` means the row belongs to that city.
- `0` means the row does not belong to that city.


## 5. Get New Encoded Column Names

One-Hot Encoding creates one new column for each category.

`get_feature_names_out()` gives the names of those newly created columns.


```python
# get all the column name
ohe.get_feature_names_out()
```

## 6. Convert Encoded Output Into DataFrame

The encoded output is not directly a normal pandas DataFrame.

So we convert it into a DataFrame and give it proper column names.

This makes the encoded result easier to read and combine with the original data.


```python
#  creating a dataframe
encoded_df = pd.DataFrame(encoded.toarray(), columns=ohe.get_feature_names_out())
encoded_df
```

### Encoded DataFrame Observation

The encoded DataFrame is easier to understand because each column has a clear name.

This also helps when we combine it with the original dataset.


## 7. Combine Original Data With Encoded Data

Now we join the original dataset and the encoded columns together.

After this step, the data contains both:

- original columns
- new numeric encoded columns


```python
#  concatinating with original dataframe
data = pd.concat([data, encoded_df], axis=1)
data
```

### Combined Data Observation

The dataset is now ready for machine learning because the city information is available in numeric form.

In real projects, we usually remove the original text column after encoding before model training.


## 8. Handle Unseen Categories

An unseen category means a value that was not present during training.

Example: the encoder learned only `Delhi`, `Mumbai`, and `Kolkata`, but later it receives `Chennai`.

Because we used `handle_unknown="ignore"`, the encoder does not give an error.

Instead, it returns all zeros for the city columns.


```python
# handling unseen category
ohe.transform([["Chennai"]]).toarray()
#  this will throw an error as Chennai is not in our dataset , so make this as unknown category and ignore
```

### Unseen Category Observation

`Chennai` was not present when the encoder was trained.

Because unknown categories are ignored, the encoded result becomes all zeros.

This is useful when test data or future data may contain new categories.


## Final Conclusion

One-Hot Encoding is used for nominal categorical columns.

It is best when categories do not have any natural order.

In this notebook, we encoded the `City` column.

Important points:

1. Each unique category becomes a new column.
2. Encoded values are only `0` and `1`.
3. `handle_unknown="ignore"` helps avoid errors for new unseen categories.
4. One-Hot Encoding is simple and useful for columns with a small number of categories.
5. If a column has too many unique values, One-Hot Encoding can create too many columns.

So, One-Hot Encoding is a good choice for small nominal categorical features like `City`.
