# Label Encoding

Label Encoding is used to convert categorical labels into numbers.

Example:

- `Failed` can become `0`
- `Passed` can become `1`

## Very Important Point

Label Encoding should mainly be used for the **target variable**.

It should not usually be used for input features because it can create a fake order between categories.

For example, if we encode cities like this:

- Delhi = 0
- Kolkata = 1
- Mumbai = 2

A model may think Mumbai is greater than Kolkata and Kolkata is greater than Delhi. But city names do not have this kind of order.

For nominal input features, use One-Hot Encoding instead.

## What We Will Learn

1. Import required libraries
2. Create a simple target column
3. Apply Label Encoding
4. View the encoded result


## 1. Import Required Libraries

We import pandas for creating the DataFrame.

We import `LabelEncoder` from scikit-learn to convert target labels into numbers.


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

## 2. Create a Simple Target Column

Here we create a small DataFrame with one column called `Result`.

This column has two target labels:

- `Passed`
- `Failed`

This is a good example for Label Encoding because `Result` is the output or target variable.


```python
#  Label encoding is a technique used to convert categorical variables into numerical variables.
#  it is used for target variable
df = pd.DataFrame({"Result": ["Passed", "Failed", "Passed", "Failed", "Passed"]})
df
```

### Dataset Observation

The `Result` column contains text labels.

Machine learning models need target labels in numeric form, so we encode this column.


## 3. Apply Label Encoding

Now we create a `LabelEncoder` object.

Then we use `fit_transform()` on the `Result` column.

`fit_transform()` does two things:

1. It learns the unique labels.
2. It converts each label into a number.


```python
#  create a encoder
label_encoder = LabelEncoder()
df["Result_encoded"] = label_encoder.fit_transform(df["Result"])
```

### Encoding Observation

The new column `Result_encoded` stores the numeric form of the target labels.

The original `Result` column is still kept here so we can compare both values easily.


## 4. View the Encoded Data

Now we display the DataFrame after adding the encoded target column.


```python
df
```

### Final Data Observation

The target variable is now available in numeric form.

This encoded target can be used while training a classification model.


## Final Conclusion

Label Encoding converts target labels into numbers.

In this notebook, we encoded the `Result` target column.

Important points:

1. Label Encoding is mainly used for the **target variable**.
2. Do not use Label Encoding for nominal input features like city, color, or country.
3. Label Encoding can create a false order between categories.
4. For nominal input features, One-Hot Encoding is usually a better choice.
5. Label Encoding is useful for target labels in classification problems.
