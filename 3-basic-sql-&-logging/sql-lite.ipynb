```python
import sqlite3
```

```python
# creating a connection object
conn = sqlite3.connect("test.db")
# creating a cursor object
cursor = conn.cursor()
```

```python
# Create a table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS employees(
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            age INTEGER NOT NULL,
            department TEXT
        )
 """)
```

```python
# Insert data into the table
cursor.execute("""
    INSERT INTO employees(name,age,department)
        VALUES
            ('JOHN',23,'IT'),
            ('RAHUL',23,'IT'),
            ('PURSOTAM',23,'IT'),
            ('ROHAN',23,'IT'),
            ('CENA',23,'IT');
""")
#  This will make sure the changes are saved to the database
conn.commit()
```

```python
# Retrieve data from the table
cursor.execute("SELECT * FROM employees")
row = cursor.fetchone()
print(row)
```

```python
cursor.execute("""
    UPDATE employees
        SET age = 24, department = 'CSE'
        WHERE name ="JOHN";
 """)
conn.commit()
```

```python
# Delete data from the table
cursor.execute("DELETE FROM employees WHERE name = 'JOHN'")
conn.commit()
```

```python
# close the connection
conn.close()
```
