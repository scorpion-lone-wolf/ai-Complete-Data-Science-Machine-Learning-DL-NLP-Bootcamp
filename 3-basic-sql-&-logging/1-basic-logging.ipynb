```python
import logging
```

```python
# level act like a filter
# any log message whose level is less than then configured level will be ignored
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s | %(levelname)s | %(message)s",  # here s means string
    filename="app.log",
    filemode="a",
)
logging.debug("This is debug message")
logging.info("This is info message")
logging.warning("This is warning message")
logging.error("This is error message")
logging.critical("This is critical message")
```
